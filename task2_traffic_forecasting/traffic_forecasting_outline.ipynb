# Task 2 — Dự báo lưu lượng xe

## Dữ liệu
Mô tả dữ liệu Metro Interstate Traffic Volume và các đặc trưng đầu vào; cột mục tiêu là `traffic_volume`.

## Tiền xử lý
Trình bày cách xử lý thời gian, giá trị thiếu, mã hóa đặc trưng, chuẩn hóa và tạo chuỗi đầu vào/đầu ra.

## Mô hình
Mô tả LSTM Encoder–Decoder với Attention và cách tạo dự báo tại 5 mốc liên tiếp.

## Huấn luyện
Ghi cấu hình huấn luyện, hàm mất mát, optimizer và cách chọn checkpoint.

## Đánh giá
Nêu cách chia tập dữ liệu, các chỉ số đánh giá và kết quả.
