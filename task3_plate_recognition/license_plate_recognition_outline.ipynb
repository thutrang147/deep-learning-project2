# Task 3 — Nhận dạng biển số

## Dữ liệu
Mô tả tập Taiwan License Plate Recognition, cấu trúc ảnh và nhãn `license_number`.

## Tiền xử lý
Trình bày cách đọc `xywhr`, cắt vùng biển số, biến đổi ảnh và mã hóa chuỗi ký tự.

## Mô hình
Mô tả CNN Encoder và LSTM Decoder dùng để nhận dạng chuỗi ký tự.

## Huấn luyện
Ghi cấu hình huấn luyện, hàm mất mát, optimizer và cách chọn checkpoint.

## Đánh giá
Nêu cách chia tập dữ liệu, các chỉ số đánh giá và kết quả.
